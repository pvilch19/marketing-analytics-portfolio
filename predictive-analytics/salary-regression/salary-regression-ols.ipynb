# Simple Linear Regression

```python
import pandas as pd
import statsmodels.formula.api as smf
```

```python
# Import Salary_Data
dataset = pd.read_csv('data/Salary_Data.csv')


# Specify and estimate the regression model (IV: YearsExperience , DV: Salary)
model = smf.ols("Salary ~ YearsExperience", data=dataset).fit()
```

```python
# Summarize the results
print(model.summary())
```

Interpret the fitted relationship as an association within this small teaching dataset. A high R-squared does not establish causation or reliable predictions across occupations and labor markets. A displayed p-value of 0.000 is rounded, not literally zero.
